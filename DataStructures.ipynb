1 What are data structures, and why are they important

Data structures are ways of organizing and storing data in a computer so that it can be accessed and modified efficiently.

They are important because they help us manage data, perform operations easily, and improve program efficiency.
     

2 Explain the difference between mutable and immutable data types with examples

Mutable data types are data types whose values can be changed after creation, whereas immutable data types cannot be changed after creation.


3 What are the main differences between lists and tuples in Python

Lists and tuples are both used to store multiple elements in Python. The main difference is that lists are mutable, whereas tuples are immutable. Lists are created using square brackets [], while tuples are created using parentheses ()

4 Describe how dictionaries store data

In Python, a dictionary stores data in key-value pairs. Each key is unique and is used to access its corresponding value. Dictionaries are created using curly braces {}.


5  Why might you use a set instead of a list in Python

We might use a set instead of a list in Python when we want to store only unique elements and remove duplicates. A set automatically removes duplicate values

6  What is a string in Python, and how is it different from a list

A string in Python is a sequence of characters enclosed in single or double quotes. A list is a collection of elements enclosed in square brackets []. The main difference is that strings are immutable, whereas lists are mutable.



7  How do tuples ensure data integrity in Python

Tuples ensure data integrity in Python because they are immutable, meaning their elements cannot be changed after creation. This prevents accidental modification of data and helps keep it consistent and reliable throughout a program.

8 What is a hash table, and how does it relate to dictionaries in Python

A hash table is a data structure that stores data as key-value pairs and allows values to be accessed quickly using keys. In Python, dictionaries use a hash table internally to store and retrieve values efficiently.

9 Can lists contain different data types in Python

Yes, lists in Python can contain different data types in the same list. For example, a list can store integers, strings, floats, and Boolean values together.

10 Explain why strings are immutable in Python

Strings are immutable in Python, which means their characters cannot be changed after the string is created. This helps maintain data consistency and makes strings safer to use in programs


11 What advantages do dictionaries offer over lists for certain tasks

Dictionaries offer several advantages over lists when we need to store and retrieve data using keys instead of numerical indexes. They allow us to access values directly using meaningful keys, making the code easier to read and understand. Dictionaries also provide efficient searching by key and are useful for storing related information as key-value pairs.

12 Describe a scenario where using a tuple would be preferable over a list

Tuples are preferable when data should remain constant. For example, storing the coordinates of a location in a tuple ensures that the values cannot be accidentally modified.

13 How do sets handle duplicate values in Python

Sets automatically eliminate duplicate values and store only unique elements. They are useful when we need to remove duplicates from a collection of data

14 How does the “in” keyword work differently for lists and dictionaries

In lists, the in keyword checks whether a value is present in the list. In dictionaries, it checks whether a key exists by default, not whether a value is present.

15 Can you modify the elements of a tuple? Explain why or why not

No, tuple elements cannot be modified after creation because tuples are immutable. This helps protect data from accidental changes and maintain data integrity.

16 What is a nested dictionary, and give an example of its use case

A nested dictionary is a dictionary that contains another dictionary as its value. It is useful for organizing related information in a structured way.

17 Describe the time complexity of accessing elements in a dictionary

Dictionary access has an average time complexity of O(1) because Python uses hashing to retrieve values quickly by key. In the worst case, it can be O(n).


18  In what situations are lists preferred over dictionaries

Lists are preferred over dictionaries when we need to store an ordered collection of items and access them using numerical indexes rather than keys. Lists are useful when duplicate values are allowed, when we need to maintain items in sequence, or when we want to add, remove, or modify elements easily.

19 Why are dictionaries considered unordered, and how does that affect data retrieval

Dictionaries were traditionally considered unordered because their items were not guaranteed to maintain insertion order. In Python 3.7 and later, dictionaries preserve insertion order. Data retrieval is still efficient because values are accessed using keys rather than indexes.

20 Explain the difference between a list and a dictionary in terms of data retrieval.

Lists access elements using numerical indexes, whereas dictionaries access values using keys. Both allow efficient data retrieval, but dictionaries are especially useful when we need to find information using a specific key.
